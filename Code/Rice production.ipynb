```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. โหลดข้อมูล
df = pd.read_csv(r"C:\Users\Sasitorn Pansorn\Downloads\Project ไพธอน\Rice production.csv")

# 2. เลือกปีล่าสุดที่มีข้อมูล (เช่น 2021)
latest_year = df['Year'].max()
df_latest = df[df['Year'] == latest_year].dropna(subset=['Value'])

# 3. กรองเลือกเฉพาะประเทศหลักๆ (ตัดกลุ่มภูมิภาคใหญ่ที่อาจทำให้ค่าซ้ำออก เช่น World หรือรวมกลุ่ม)
# ในตัวอย่างนี้เราเลือกเฉพาะแถวที่มีค่า Value มากที่สุด 5 อันดับแรกโดยตรง
top_5 = df_latest.nlargest(5, 'Value')

# คำนวณยอดรวมของประเทศอื่นๆ นอกเหนือจาก Top 5
others_sum = df_latest[~df_latest['Area'].isin(top_5['Area'])]['Value'].sum()

# 4. เตรียมข้อมูลสำหรับ Pie Chart
labels = list(top_5['Area']) + ['Others']
values = list(top_5['Value']) + [others_sum]

# 5. วาดกราฟ
plt.figure(figsize=(9, 9))
plt.pie(
    values,
    labels=labels,
    autopct='%1.1f%%',
    startangle=140,
    colors=sns.color_palette('pastel'),
    wedgeprops={'edgecolor': 'white', 'linewidth': 1},
)

plt.title(
    f'Share of Global Rice Production (Top 5 + Others) - {latest_year}',
    fontsize=14,
    fontweight='bold',
    pad=20,
)
plt.tight_layout()
plt.show()
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ปรับชื่อไฟล์ตรงนี้ให้ตรงกับของคุณ (ลองให้ทั้งชื่อมีขีดล่างและเว้นวรรค)
CSV_PATH = next((p for p in ["Rice_production.csv", "Rice production.csv"] if os.path.exists(p)),
                "Rice_production.csv")
OUT_DIR = "outputs"
os.makedirs(OUT_DIR, exist_ok=True)

BLUE, PINK, PURPLE, CYAN, ORANGE = "#3b4cc0", "#ec4899", "#8b5cf6", "#22b8d4", "#f9a825"
PALETTE = [BLUE, PURPLE, CYAN, PINK, ORANGE, "#10b981", "#64748b"]


def load_rice(path=CSV_PATH):
    df = pd.read_csv(path)
    # ตัดยอดรวม/ประเทศเดิมที่ซ้ำซ้อน แล้วใช้ 'China, mainland' แทน China
    df = df[~df["Area"].isin(["China", "USSR", "Yugoslav SFR", "Sudan (former)"])].copy()
    df["Area"] = df["Area"].replace({"China, mainland": "China"})
    df["Mt"] = df["Value"] / 1e6  # ล้านตัน
    return df.reset_index(drop=True)


def wide(df):
    return df.pivot_table(index="Year", columns="Area", values="Mt")


def top_areas(df, year=2021, n=5):
    return df[df["Year"] == year].nlargest(n, "Mt")["Area"].tolist()

df = load_rice()
W = wide(df)
print(df.shape, "OK")

import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "font.sans-serif": ["Leelawadee UI", "Tahoma", "Noto Sans Thai", "Loma", "DejaVu Sans"],
    "axes.unicode_minus": False,
})


# ------------------------------------------------------------ เตรียมข้อมูลสำหรับ regression
# ตัวอย่างเดิมคือ import vs export ของแต่ละประเทศ -> ข้อมูลข้าวมีตัวแปรเดียว
# จึงใช้ x = ผลผลิตเฉลี่ย 1961-1990, y = ผลผลิตเฉลี่ย 1991-2021 ของแต่ละประเทศ (พันตัน)
early = W.loc[1961:1990].mean()
late = W.loc[1991:2021].mean()
d = (early.to_frame("x").join(late.to_frame("y")) / 1e-3).dropna()
d = d[(d.x > 0) & (d.y > 0)].sort_values("x")
x, y = d["x"].to_numpy(), d["y"].to_numpy()

# ------------------------------------------------------------ ภาพที่13 Linear Regression
m, c = np.polyfit(x, y, 1)
fig, ax = plt.subplots(figsize=(6.5, 5))
ax.scatter(x, y, color="green", label="rawdata")
ax.plot(x, m * x + c, color="black", label="regression line")
ax.set(title="Linear Regression", xlabel="production 1961-1990 (thousand tonnes)",
       ylabel="production 1991-2021 (thousand tonnes)")
ax.legend(); fig.tight_layout()
fig.savefig(f"{OUT_DIR}/13_linear_regression.png", dpi=150); plt.close(fig)
r2 = np.corrcoef(x, y)[0, 1] ** 2
print(f"linear: y={m:.3f}x+{c:.1f}  R2={r2:.3f}")

# ------------------------------------------------------------ ภาพที่14 Polynomial Regression
deg = 3
poly = np.poly1d(np.polyfit(x, y, deg))
xs = np.linspace(x.min(), x.max(), 300)
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(x, y, color="green", label="raw data")
ax.plot(xs, poly(xs), color="gray", label="Polynomial Regression")
ax.set(title="relationship btw production 1961-1990 and 1991-2021 (rice)",
       xlabel="production 1961-1990 (thousand tonnes)",
       ylabel="production 1991-2021 (thousand tonnes)")
ax.grid(True); ax.legend(); fig.tight_layout()
fig.savefig(f"{OUT_DIR}/14_polynomial_regression.png", dpi=150); plt.close(fig)

# ------------------------------------------------------------ ภาพที่7-10 Line Graph หลายประเทศ
def line_graph(areas, title, fname, y0=2017, y1=2021, legend_loc="best"):
    t = W.loc[y0:y1, areas]
    fig, ax = plt.subplots(figsize=(7, 4.5))
    for a in areas:
        ax.plot(t.index, t[a] / 1e-3, marker=".", label=f"Area {a}")   # พันตัน
    ax.set_xticks(t.index)                       # ปีเป็นจำนวนเต็ม (ตัวอย่างเดิมเป็น 2018.5 ฯลฯ)
    ax.tick_params(axis="x", rotation=90)
    ax.set(title=title, xlabel="Year", ylabel="Value (thousand tonnes)")
    ax.legend(loc=legend_loc, fontsize=8); fig.tight_layout()
    fig.savefig(f"{OUT_DIR}/{fname}.png", dpi=150); plt.close(fig)

ranked = df[df["Year"] == 2021].sort_values("Mt", ascending=False)["Area"].tolist()
line_graph(ranked[:5], "Rice Production: Top 5 Producers", "07_line_top5", legend_loc="center left")
line_graph(ranked[5:10], "Rice Production: Producers Ranked 6-10", "08_line_rank6_10")
# กลุ่มประเทศเพื่อนบ้านไทย (คล้ายตัวอย่างที่เลือก 5 ประเทศใกล้เคียง)
line_graph(["Thailand", "Viet Nam", "Myanmar", "Cambodia", "Lao People's Democratic Republic"],
           "Rice Production: Thailand and Neighbours", "09_line_neighbours")
print("part3 done")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ปรับชื่อไฟล์ตรงนี้ให้ตรงกับของคุณ (ลองให้ทั้งชื่อมีขีดล่างและเว้นวรรค)
CSV_PATH = next((p for p in ["Rice_production.csv", "Rice production.csv"] if os.path.exists(p)),
                "Rice_production.csv")
OUT_DIR = "outputs"
os.makedirs(OUT_DIR, exist_ok=True)

BLUE, PINK, PURPLE, CYAN, ORANGE = "#3b4cc0", "#ec4899", "#8b5cf6", "#22b8d4", "#f9a825"
PALETTE = [BLUE, PURPLE, CYAN, PINK, ORANGE, "#10b981", "#64748b"]


def load_rice(path=CSV_PATH):
    df = pd.read_csv(path)
    # ตัดยอดรวม/ประเทศเดิมที่ซ้ำซ้อน แล้วใช้ 'China, mainland' แทน China
    df = df[~df["Area"].isin(["China", "USSR", "Yugoslav SFR", "Sudan (former)"])].copy()
    df["Area"] = df["Area"].replace({"China, mainland": "China"})
    df["Mt"] = df["Value"] / 1e6  # ล้านตัน
    return df.reset_index(drop=True)


def wide(df):
    return df.pivot_table(index="Year", columns="Area", values="Mt")


def top_areas(df, year=2021, n=5):
    return df[df["Year"] == year].nlargest(n, "Mt")["Area"].tolist()

df = load_rice()
W = wide(df)
print(df.shape, "OK")

import time
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "font.family": "sans-serif",
    "font.sans-serif": ["Leelawadee UI", "Tahoma", "Noto Sans Thai", "Loma", "DejaVu Sans"],
    "axes.unicode_minus": False,
})


def style(ax, grid_axis="both"):
    ax.spines[["top", "right"]].set_visible(False)
    ax.grid(True, axis=grid_axis, ls=":", alpha=.5)
    ax.set_axisbelow(True)



# ---------------------------------------------------------------- 1) VECTORIZATION
# a = ผลผลิตข้าว (ตัน) สุ่มตัวอย่างจากข้อมูลจริง, b = ราคาสมมติ (USD/ตัน) -> รายได้ = a*b
rng = np.random.default_rng(0)
pool = df["Value"].to_numpy()
sizes = [10_000, 100_000, 1_000_000]
t_loop, t_np = [], []
for n in sizes:
    a = rng.choice(pool, n)
    b = rng.uniform(300, 500, n)
    best_loop = best_np = float("inf")
    for _ in range(3):
        t0 = time.perf_counter(); c = [a[i] * b[i] for i in range(len(a))]
        best_loop = min(best_loop, time.perf_counter() - t0)
        t0 = time.perf_counter(); c = a * b
        best_np = min(best_np, time.perf_counter() - t0)
    t_loop.append(best_loop * 1000); t_np.append(best_np * 1000)

fig, ax = plt.subplots(figsize=(6.2, 4.2))
x = np.arange(len(sizes)); w = .38
b1 = ax.bar(x - w/2, t_loop, w, color=PINK, label="for loop")
b2 = ax.bar(x + w/2, t_np, w, color=BLUE, label="NumPy vectorized")
ax.set_yscale("log")
ax.set_xticks(x, [f"{n:,}" for n in sizes])
ax.set_xlabel("จำนวนข้อมูล (แถว)"); ax.set_ylabel("เวลา (มิลลิวินาที, log)")
ax.bar_label(b1, fmt="%.2f", fontsize=8); ax.bar_label(b2, fmt="%.2f", fontsize=8)
ax.set_title("for loop vs NumPy (ข้อมูลผลผลิตข้าว)", color=BLUE, fontweight="bold")
ax.legend(); style(ax, "y")
fig.tight_layout(); fig.savefig(f"{OUT_DIR}/01_vectorization.png", dpi=160); plt.close(fig)

# ---------------------------------------------------------------- 2) STATISTICS
th = W["Thailand"].dropna()
xs, ys = th.index.to_numpy(float), th.to_numpy()
m, b0 = np.polyfit(xs, ys, 1)                 # เส้นตรงกำลังสองน้อยที่สุด
r = np.corrcoef(xs, ys)[0, 1]                 # สหสัมพันธ์
# การกระจายผลผลิตข้ามประเทศ ปี 2021 (log10 เพราะเบ้ขวามาก)
v21 = df[df["Year"] == 2021]["Value"].to_numpy()
lv = np.log10(v21[v21 > 0])
q1, q2, q3 = np.percentile(lv, [25, 50, 75])  # ควอไทล์

fig, (a1, a2) = plt.subplots(1, 2, figsize=(10.5, 3.8))
a1.scatter(xs, ys, s=14, color=PURPLE, alpha=.7)
a1.plot(xs, m * xs + b0, color=BLUE, lw=2.2, label=f"y={m:.3f}x{b0:+.1f}")
a1.set_title(f"np.polyfit | r = {r:.3f}  (ไทย)", color=BLUE, fontweight="bold")
a1.set_xlabel("ปี"); a1.set_ylabel("ผลผลิต (ล้านตัน)"); a1.legend(); style(a1)

a2.hist(lv, bins=20, color="#c7d2fe", edgecolor="white")
for q, lab, c in [(q1, "Q1", CYAN), (q2, "Median", PINK), (q3, "Q3", ORANGE)]:
    a2.axvline(q, color=c, ls="--", lw=2, label=f"{lab}={10**q/1e3:,.0f} พันตัน")
a2.set_title("np.percentile (ปี 2021)", color=BLUE, fontweight="bold")
a2.set_xlabel("log10(ผลผลิต, ตัน)"); a2.set_ylabel("จำนวนประเทศ"); a2.legend(fontsize=8); style(a2)
fig.tight_layout(); fig.savefig(f"{OUT_DIR}/02_statistics.png", dpi=160); plt.close(fig)
print(f"polyfit: m={m:.4f}, b={b0:.3f}, r={r:.4f}, percentiles(tonnes)={10**np.array([q1,q2,q3])}")

# ---------------------------------------------------------------- 3) NUMPY + MATPLOTLIB
years = np.arange(1961, 2022)
raw = W["Thailand"].reindex(years).to_numpy()
k = 5
ma = np.convolve(raw, np.ones(k) / k, mode="valid")   # ค่าเฉลี่ยเคลื่อนที่ 5 ปี
yma = years[k // 2: -(k // 2)]

fig, ax = plt.subplots(figsize=(6.5, 4))
ax.plot(years, raw, color=BLUE, lw=2.2, label="ผลผลิตรายปี")
ax.plot(yma, ma, "--", color=PINK, lw=2.2, label="เฉลี่ยเคลื่อนที่ 5 ปี")
ax.fill_between(yma, raw[k // 2: -(k // 2)], ma, color=PURPLE, alpha=.15)
ax.set_ylabel("ล้านตัน"); ax.set_title("ผลผลิตข้าวไทย: NumPy + Matplotlib", color=BLUE, fontweight="bold")
ax.legend(loc="lower right"); style(ax)
fig.tight_layout(); fig.savefig(f"{OUT_DIR}/03_numpy_matplotlib.png", dpi=160); plt.close(fig)
print("part1 done")
```

```python
"""OVERFITTING: เทียบ train RMSE กับ test RMSE ตามดีกรีของ polynomial (ข้อมูลข้าวเวียดนาม)
ไฟล์เดียวจบ วางในเซลล์ Jupyter ได้เลย | ต้องมี numpy, pandas, matplotlib (plotly ไม่บังคับ)"""
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---------------- ตั้งค่า ----------------
CSV_PATH = next((p for p in ["Rice_production.csv", "Rice production.csv"] if os.path.exists(p)),
                "Rice_production.csv")
AREA = "Viet Nam"                               # เปลี่ยนเป็นประเทศอื่นได้ เช่น "Indonesia"
DEGREES = [1, 2, 3, 4, 5, 6, 8, 10, 12, 15]
N_SPLITS = 100                                  # สุ่มแบ่ง train/test ซ้ำ แล้วใช้ค่ามัธยฐานให้กราฟนิ่ง
BLUE, PINK, PURPLE = "#3b4cc0", "#ec4899", "#8b5cf6"
plt.rcParams.update({
    "font.sans-serif": ["Leelawadee UI", "Tahoma", "Noto Sans Thai", "Loma", "DejaVu Sans"],
    "axes.unicode_minus": False,
})

# ---------------- เตรียมข้อมูล ----------------
df = pd.read_csv(CSV_PATH)
s = df[df["Area"] == AREA].set_index("Year")["Value"] / 1e6        # ล้านตัน
x = ((s.index - s.index.min()) / (s.index.max() - s.index.min()) * 2 - 1).to_numpy()  # สเกลปี -> [-1, 1]
y = s.to_numpy()

# ---------------- ฟิตโมเดลแต่ละดีกรี ----------------
rmse = lambda a, b: float(np.sqrt(np.mean((a - b) ** 2)))
tr_all = {d: [] for d in DEGREES}
te_all = {d: [] for d in DEGREES}
for seed in range(N_SPLITS):
    idx = np.random.default_rng(seed).permutation(len(x))
    k = int(0.7 * len(x))
    tr, te = idx[:k], idx[k:]
    for d in DEGREES:
        p = np.polynomial.Polynomial.fit(x[tr], y[tr], d)   # เสถียรกว่า np.polyfit เมื่อดีกรีสูง
        tr_all[d].append(rmse(p(x[tr]), y[tr]))
        te_all[d].append(rmse(p(x[te]), y[te]))
train = np.array([np.median(tr_all[d]) for d in DEGREES])
test = np.array([np.median(te_all[d]) for d in DEGREES])

best = int(np.argmin(test))                       # ดีกรีที่ test error ต่ำสุด = sweet spot
lo, hi = DEGREES[max(best - 1, 0)], DEGREES[min(best + 1, len(DEGREES) - 1)]
print(pd.DataFrame({"degree": DEGREES, "train RMSE": train.round(2), "test RMSE": test.round(2)}).to_string(index=False))
print(f"\nUnderfitting  degree 1 : train/test = {train[0]:.2f} / {test[0]:.2f}")
print(f"พอดี          degree {DEGREES[best]} : train/test = {train[best]:.2f} / {test[best]:.2f}")
print(f"Overfitting   degree {DEGREES[-1]}: train/test = {train[-1]:.2f} / {test[-1]:.2f}")

# ---------------- กราฟ (Matplotlib) ----------------
fig, ax = plt.subplots(figsize=(7, 4.6))
ax.axvspan(lo - .5, hi + .5, color=PURPLE, alpha=.10)
ax.plot(DEGREES, train, "-o", color=BLUE, ms=4, lw=1.6, label="train RMSE")
ax.plot(DEGREES, test, "-D", color=PINK, ms=4, lw=1.6, label="test RMSE")
ymax = max(test.max(), train.max())
ax.text((lo + hi) / 2, ax.get_ylim()[0] + .02 * ymax, "sweet spot", ha="center", va="bottom", fontsize=9, color="#333")
ax.text(DEGREES[-3], ax.get_ylim()[0] + .02 * ymax, "overfitting", ha="center", va="bottom", fontsize=9, color="#333")
ax.set(xlabel="polynomial degree", ylabel="RMSE, million tonnes (lower is better)",
       title=f"Train vs Test error ({AREA})")
ax.set_xticks([2, 4, 6, 8, 10, 12, 14])
ax.spines[["top", "right"]].set_visible(False)
ax.grid(True, color="#e5e7eb"); ax.set_axisbelow(True)
ax.legend(frameon=False, loc="upper right")
fig.tight_layout()
os.makedirs("outputs", exist_ok=True)
fig.savefig("outputs/overfitting.png", dpi=160)
plt.show()

# ---------------- กราฟ (Plotly, สไตล์เดียวกับตัวอย่าง) ----------------
try:
    import plotly.graph_objects as go
    f = go.Figure()
    f.add_vrect(x0=lo - .5, x1=hi + .5, fillcolor=PURPLE, opacity=.10, line_width=0)
    f.add_trace(go.Scatter(x=DEGREES, y=train, mode="lines+markers", name="train RMSE",
                           line=dict(color=BLUE), marker=dict(symbol="circle", size=6)))
    f.add_trace(go.Scatter(x=DEGREES, y=test, mode="lines+markers", name="test RMSE",
                           line=dict(color=PINK), marker=dict(symbol="diamond", size=7)))
    f.add_annotation(x=(lo + hi) / 2, y=0, yref="paper", text="sweet spot", showarrow=False, yanchor="bottom")
    f.add_annotation(x=DEGREES[-3], y=0, yref="paper", text="overfitting", showarrow=False, yanchor="bottom")
    f.update_layout(template="plotly_white", xaxis_title="polynomial degree",
                    yaxis_title="RMSE (lower is better)", title=f"Train vs Test error ({AREA})",
                    legend=dict(x=.98, y=.98, xanchor="right"))
    f.write_html("outputs/overfitting.html")
    f.show()
except ImportError:
    print("ไม่มี plotly: ข้ามกราฟ Plotly (pip install plotly)")
```

```python
"""WORKFLOW 5 ขั้นตอน: จำลองข้อมูล -> px.scatter -> PolynomialFeatures -> r2_score -> train_test_split
ไฟล์เดียวจบ วางในเซลล์ Jupyter ได้เลย | pip install numpy pandas scikit-learn matplotlib plotly"""
import os
import numpy as np
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split

# ---------------- ตั้งค่า ----------------
USE_RICE_CSV = False   # False = จำลองปุ๋ย-ผลผลิตตามสไลด์ | True = ใช้ปี vs ผลผลิตข้าวจริงจากไฟล์ CSV
CSV_PATH = next((p for p in ["Rice_production.csv", "Rice production.csv"] if os.path.exists(p)),
                "Rice_production.csv")
AREA = "Viet Nam"
DEGREE = 3
os.makedirs("outputs", exist_ok=True)

# ================= 1) จำลองข้อมูล =================
if not USE_RICE_CSV:
    # ปริมาณปุ๋ย (x) กับผลผลิต (y) 40 แปลง: ผลผลิตเพิ่มขึ้นช่วงแรกแล้วลดลงเมื่อใส่ปุ๋ยมากเกินไป
    rng = np.random.default_rng(42)
    x = np.sort(rng.uniform(0, 10, 40))
    y = -0.45 * (x - 6) ** 2 + 18 + rng.normal(0, 1.2, 40)
    df = pd.DataFrame({"fertilizer": x, "yield_": y})
    XCOL, YCOL = "fertilizer", "yield_"
    XLAB, YLAB = "ปริมาณปุ๋ย (x)", "ผลผลิต (y)"
else:
    d = pd.read_csv(CSV_PATH)
    d = d[d["Area"] == AREA].sort_values("Year")
    df = pd.DataFrame({"year": d["Year"].to_numpy(), "production": d["Value"].to_numpy() / 1e6})
    XCOL, YCOL, XLAB, YLAB = "year", "production", "ปี", "ผลผลิตข้าว (ล้านตัน)"
print(df.head(), f"\n{len(df)} แถว\n")

X = df[[XCOL]].to_numpy()
y = df[YCOL].to_numpy()
mu, sd = X.mean(), X.std()          # สเกล x ให้เสถียร (สำคัญมากเมื่อ x เป็นปี เช่น 1961-2021)
Xs = (X - mu) / sd

# ================= 2) วาดกราฟ px.scatter =================
try:
    import plotly.express as px
    fig = px.scatter(df, x=XCOL, y=YCOL, title="ความสัมพันธ์ระหว่าง x และ y (px.scatter)",
                     labels={XCOL: XLAB, YCOL: YLAB}, color_discrete_sequence=["#3b4cc0"])
    fig.update_layout(template="plotly_white")
    fig.show()
except ImportError:
    import matplotlib.pyplot as plt
    plt.scatter(df[XCOL], df[YCOL], color="#3b4cc0"); plt.xlabel(XLAB); plt.ylabel(YLAB)
    plt.title("ความสัมพันธ์ระหว่าง x และ y"); plt.show()

# ================= 3) fit พหุนาม (PolynomialFeatures) =================
poly = PolynomialFeatures(degree=DEGREE, include_bias=False)
X_poly = poly.fit_transform(Xs)                  # [x, x^2, x^3]
model = LinearRegression().fit(X_poly, y)
y_hat = model.predict(X_poly)
print(f"ดีกรี {DEGREE}: intercept = {model.intercept_:.3f}, coef = {np.round(model.coef_, 3)}")

# ================= 4) หา R² (r2_score) =================
print(f"R² (ทั้งชุดข้อมูล) = {r2_score(y, y_hat):.4f}")

# ================= 5) แบ่ง train / test (train_test_split) =================
Xtr, Xte, ytr, yte = train_test_split(Xs, y, test_size=0.3, random_state=42)
rows = []
for deg in [1, 2, 3, 5, 8, 12]:
    p = PolynomialFeatures(degree=deg, include_bias=False)
    m = LinearRegression().fit(p.fit_transform(Xtr), ytr)
    rows.append({"degree": deg,
                 "R² train": r2_score(ytr, m.predict(p.transform(Xtr))),
                 "R² test": r2_score(yte, m.predict(p.transform(Xte)))})
res = pd.DataFrame(rows).round(3)
print("\nเทียบ train / test ตามดีกรี\n", res.to_string(index=False))

# ---------------- กราฟสรุป: ข้อมูล + เส้นพหุนาม + จุด train/test ----------------
import matplotlib.pyplot as plt
plt.rcParams.update({"font.sans-serif": ["Leelawadee UI", "Tahoma", "Noto Sans Thai", "Loma", "DejaVu Sans"],
                     "axes.unicode_minus": False})
grid = np.linspace(Xs.min(), Xs.max(), 300).reshape(-1, 1)
m3 = LinearRegression().fit(poly.fit_transform(Xtr), ytr)           # โมเดลที่ฟิตด้วย train เท่านั้น
fig, ax = plt.subplots(figsize=(7, 4.6))
ax.scatter(Xtr * sd + mu, ytr, color="#3b4cc0", label="train")
ax.scatter(Xte * sd + mu, yte, color="#ec4899", marker="D", label="test")
ax.plot(grid * sd + mu, m3.predict(poly.transform(grid)), color="#8b5cf6", lw=2.2,
        label=f"poly degree {DEGREE}  (R² test = {r2_score(yte, m3.predict(poly.transform(Xte))):.2f})")
ax.set(xlabel=XLAB, ylabel=YLAB, title="Polynomial regression + train/test split")
ax.spines[["top", "right"]].set_visible(False); ax.grid(True, color="#e5e7eb"); ax.legend(frameon=False)
fig.tight_layout(); fig.savefig("outputs/workflow_polynomial.png", dpi=160); plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Tahoma"   # ฟอนต์ไทย

df = pd.read_csv("Rice production.csv")  # แก้เป็น path ของคุณ

# 1) คำนวณความชัน, R² และจำนวนเท่าของแต่ละประเทศ
countries = {"Indonesia": "อินโดนีเซีย", "Viet Nam": "เวียดนาม", "Thailand": "ไทย"}
rows = []
for eng, thai in countries.items():
    d = df[df["Area"] == eng].sort_values("Year")
    x = d["Year"].values
    y = d["Value"].values / 1e6                    # แปลงเป็นล้านตัน
    slope, intercept = np.polyfit(x, y, 1)         # เส้นถดถอยเชิงเส้น
    r2 = np.corrcoef(x, y)[0, 1] ** 2
    rows.append((thai, slope, r2, y[-1] / y[0]))

res = pd.DataFrame(rows, columns=["ประเทศ", "slope", "r2", "mult"])
res = res.sort_values("slope", ascending=False).reset_index(drop=True)

# 2) วาดกราฟแท่งแนวนอน (ไทยสีทอง ที่เหลือสีน้ำเงิน)
colors = ["#e3a52c" if c == "ไทย" else "#3d47c9" for c in res["ประเทศ"]]

fig, ax = plt.subplots(figsize=(11, 4.5), facecolor="#f7f6f1")
ax.set_facecolor("#f7f6f1")

y_pos = np.arange(len(res))[::-1]                  # แถวบนสุด = ความชันสูงสุด
ax.barh(y_pos, res["slope"], color=colors, height=0.5)

for yp, r in zip(y_pos, res.itertuples()):
    ax.text(r.slope + 0.02, yp, f"{r.slope:.2f} ล้านตัน/ปี",
            va="center", fontsize=13, fontweight="bold", color="#1c1f4d")
    ax.text(1.35, yp, f"R² = {r.r2:.3f}\n×{r.mult:.1f} จากปี 1961",
            va="center", ha="right", fontsize=10, color="#4a4f7a")

ax.set_yticks(y_pos)
ax.set_yticklabels(res["ประเทศ"], fontsize=14, fontweight="bold", color="#1c1f4d")
ax.set_xlim(0, 1.4)
ax.set_xlabel("ความชัน = ผลผลิตที่เพิ่มขึ้นเฉลี่ยต่อปี (ล้านตัน/ปี)", fontsize=11, color="#4a4f7a")
ax.set_title("ไทยโตช้ากว่าเวียดนามและอินโดนีเซีย", fontsize=18,
             fontweight="bold", color="#1c1f4d", loc="left")

for s in ["top", "right", "left"]:
    ax.spines[s].set_visible(False)
ax.tick_params(left=False)

plt.tight_layout()
plt.savefig("regression_slope.png", dpi=200, facecolor=fig.get_facecolor())
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import font_manager as fm
from scipy import stats

CSV_PATH = "Rice production.csv"
df = pd.read_csv(CSV_PATH)

# ---------- 1) ข้อมูล ----------
df = pd.read_csv(CSV_PATH)
countries = {"Indonesia": "อินโดนีเซีย", "Viet Nam": "เวียดนาม", "Thailand": "ไทย"}
colors    = {"Indonesia": "#1B1F4B", "Viet Nam": "#3B44C9", "Thailand": "#C98A0B"}

d = df[df["Area"].isin(countries)].copy()
d["Value"] = d["Value"] / 1e6                       # ล้านตัน

# ---------- 2) หาค่า: ความชัน, R², ค่าปีล่าสุด ----------
rows = []
for area in countries:
    s = d[d["Area"] == area].sort_values("Year")
    r = stats.linregress(s["Year"], s["Value"])
    rows.append({
        "ประเทศ": countries[area],
        "ผลผลิต 1961": round(s["Value"].iloc[0], 1),
        "ผลผลิต 2021": round(s["Value"].iloc[-1], 1),
        "ความชัน (ล้านตัน/ปี)": round(r.slope, 3),
        "R²": round(r.rvalue ** 2, 3),
    })
summary = pd.DataFrame(rows)
print(summary.to_string(index=False))

# ---------- 3) ฟอนต์ไทย ----------
available = {f.name for f in fm.fontManager.ttflist}
for name in ["Prompt", "Sarabun", "Noto Sans Thai", "Leelawadee UI", "Tahoma"]:
    if name in available:
        plt.rcParams["font.family"] = name
        break

# ---------- 4) กราฟ ----------
BG, NAVY = "#F8F6F1", "#1B1F4B"
fig, ax = plt.subplots(figsize=(8.5, 6), facecolor=BG)
ax.set_facecolor(BG)
fig.subplots_adjust(left=0.07, right=0.80, top=0.88, bottom=0.17)

for area in countries:
    s = d[d["Area"] == area].sort_values("Year")
    ax.plot(s["Year"], s["Value"], color=colors[area], lw=2.6)
    # จุดปลายเส้น + ป้ายชื่อ/ค่า
    x, y = s["Year"].iloc[-1], s["Value"].iloc[-1]
    ax.scatter([x], [y], s=110, color=colors[area], zorder=5, clip_on=False)
    ax.annotate(f"{countries[area]}\n{y:.1f}", xy=(x, y), xytext=(10, 0),
                textcoords="offset points", va="center", ha="left",
                fontsize=12, fontweight="bold", color=colors[area],
                annotation_clip=False)

ax.set_title("ผลผลิตข้าว 3 ประเทศ (ล้านตัน)", loc="left", fontsize=16,
             fontweight="bold", color=NAVY, pad=18)

ax.set_xlim(1961, 2021)
ax.set_ylim(0, 65)
ax.set_xticks([1961, 1981, 2001, 2021])
ax.set_yticks([0, 20, 40, 60])
ax.tick_params(length=0, labelsize=11, colors="#444")
ax.grid(axis="y", color="#DDD9CF", lw=1)
ax.set_axisbelow(True)
for sp in ["top", "right", "left", "bottom"]:
    ax.spines[sp].set_visible(False)

fig.text(0.07, 0.05,
         "ความชัน = ผลผลิตที่เพิ่มขึ้นเฉลี่ยต่อปีจากเส้นถดถอยเชิงเส้น · "
         "R² ใกล้ 1 = แนวโน้มใกล้เส้นตรง",
         fontsize=9.5, color="#555")

plt.savefig("rice_3countries.png", dpi=200, facecolor=BG)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# ตั้งค่าฟอนต์หลักเป็น Tahoma เพื่อรองรับภาษาไทย
plt.rcParams["font.family"] = "Tahoma"
plt.rcParams["axes.unicode_minus"] = (  # ป้องกันเครื่องหมายลบแสดงผลผิดพลาด
    False
)

# --- ข้อมูลตัวอย่างจำลอง (โครงสร้างสอดคล้องกับชุดข้อมูลจริง) ---
np.random.seed(42)
years = np.arange(1961, 2022)


def gen_data(start, slope):
  return (
      start
      + slope * (years - 1961)
      + np.sin(np.linspace(0, 15, len(years))) * 2
      + np.random.normal(0, 1, len(years))
  )


data = {
    "Year": np.tile(years, 3),
    "Area": (
        ["Thailand"] * len(years)
        + ["Vietnam"] * len(years)
        + ["Indonesia"] * len(years)
    ),
    "Value": (
        list(gen_data(10.15, 0.43))
        + list(gen_data(9.00, 0.73))
        + list(gen_data(12.08, 0.89))
    ),
}
df = pd.DataFrame(data)
# --------------------------------------------------

# 1. กรองข้อมูลเฉพาะ 3 ประเทศที่ต้องการ
countries = ["Thailand", "Vietnam", "Indonesia"]
df_filtered = df[df["Area"].isin(countries)]

# สร้างขนาดรูป (Figure Size)
fig, ax = plt.subplots(figsize=(11, 7))

# กำหนดสีของเส้นข้อมูลจริงและเส้นแนวโน้ม (Trendline)
colors = {"Thailand": "#f1948a", "Vietnam": "#85c1e9", "Indonesia": "#a9dfbf"}
trend_colors = {"Thailand": "#d9534f", "Vietnam": "#337ab7", "Indonesia": "#5cb85c"}
thai_names = {"Thailand": "ไทย", "Vietnam": "เวียดนาม", "Indonesia": "อินโดนีเซีย"}

# 2. วาดกราฟเส้นและเส้นแนวโน้ม
for country in countries:
  sub_df = df_filtered[df_filtered["Area"] == country]
  x = sub_df["Year"]
  y = sub_df["Value"]

  # เส้นกราฟข้อมูลจริง (เส้นบางโปร่งใส)
  ax.plot(
      x,
      y,
      color=colors[country],
      linewidth=1.2,
      alpha=0.85,
      solid_capstyle="round",
  )

  # คำนวณเส้นแนวโน้ม Linear Regression (Degree 1)
  z = np.polyfit(x, y, 1)
  p = np.poly1d(z)

  # เส้นประแนวโน้ม (Trendline)
  ax.plot(
      x,
      p(x),
      color=trend_colors[country],
      linestyle="--",
      linewidth=2.8,
      alpha=0.9,
  )

# 3. ตกแต่งรายละเอียดกราฟ
# ปรับแต่งแกน Y
ax.set_ylabel("ล้านตัน", fontsize=13, fontweight="bold", loc="top", rotation=0)
ax.yaxis.set_label_coords(-0.02, 1.02)  # จัดตำแหน่งคำว่า ล้านตัน ให้อยู่มุมบนซ้าย
ax.set_ylim(0, 65)
ax.set_yticks([0, 10, 20, 30, 40, 50, 60])

# เส้น Grid แนวนอนแบบบาง
ax.yaxis.grid(True, linestyle="-", alpha=0.4, color="#cccccc")
ax.set_axisbelow(True)

# ซ่อนเส้นขอบรอบกราฟ ยกเว้นแกนล่าง
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_visible(False)
ax.spines["bottom"].set_color("#aaaaaa")
ax.spines["bottom"].set_linewidth(1.2)

# 4. ใส่ข้อความชื่อประเทศกำกับไว้ใกล้ปลายเส้น (ตำแหน่งตามปีล่าสุด 2021)
```
